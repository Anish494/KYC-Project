```python
def get_face_embedding(image, upscale=3):
    if isinstance(image, Image.Image):
        # Upscale small crops so InsightFace can detect the face
        w, h = image.size
        image = image.resize((w * upscale, h * upscale), Image.LANCZOS)
        img_array = np.array(image.convert("RGB"))
        img_bgr = cv2.cvtColor(img_array, cv2.COLOR_RGB2BGR)
    else:
        img_bgr = image

    faces = face_app.get(img_bgr)

    if len(faces) == 0:
        return None, "no face detected"

    # Take the largest face
    face = max(faces, key=lambda f: (f.bbox[2]-f.bbox[0]) * (f.bbox[3]-f.bbox[1]))

    return face.embedding, "ok"


# Test again
if "photo" in detections:
    photo_crop = detections["photo"]["crop"]
    print("Original crop size:", photo_crop.size)

    emb1, status1 = get_face_embedding(photo_crop, upscale=3)
    print("Status:", status1)
    if emb1 is not None:
        print("Embedding shape:", emb1.shape)
        print("First 5 values:", emb1[:5])
```

```python
from numpy.linalg import norm

def cosine_similarity(emb1, emb2):
    return float(np.dot(emb1, emb2) / (norm(emb1) * norm(emb2)))

def match_faces(image1, image2, threshold=0.4):
    """
    Compare two face images and return match result.
    image1 - citizenship photo crop (PIL Image)
    image2 - selfie (PIL Image or file path)
    """
    # Get embedding 1
    emb1, status1 = get_face_embedding(image1)
    if emb1 is None:
        return {
            "match": False,
            "score": 0.0,
            "reason": f"citizenship photo — {status1}"
        }

    # Get embedding 2
    emb2, status2 = get_face_embedding(image2)
    if emb2 is None:
        return {
            "match": False,
            "score": 0.0,
            "reason": f"selfie — {status2}"
        }

    # Compare
    score = cosine_similarity(emb1, emb2)

    if score >= 0.6:
        confidence = "high"
    elif score >= 0.4:
        confidence = "medium"
    else:
        confidence = "low"

    return {
        "match": score >= threshold,
        "score": round(score, 4),
        "confidence": confidence,
        "reason": "ok"
    }

print("match_faces function ready")
```

```python
def verify_face(citizenship_image_path, selfie_path, threshold=0.4):
    """
    Full pipeline:
    1. YOLO detects photo field from citizenship image
    2. Extract face embedding from cropped photo
    3. Extract face embedding from selfie
    4. Compare and return result
    """

    # Step 1 — detect fields
    detections = detect_and_crop(citizenship_image_path)

    if "photo" not in detections:
        return {
            "match": False,
            "score": 0.0,
            "reason": "photo field not detected in citizenship document",
            "confidence": None
        }

    # Step 2 — get citizenship photo crop
    photo_crop = detections["photo"]["crop"]
    yolo_conf = detections["photo"]["confidence"]

    # Step 3 — load selfie
    selfie_img = Image.open(selfie_path).convert("RGB")

    # Step 4 — match
    result = match_faces(photo_crop, selfie_img, threshold=threshold)

    # Add extra info
    result["yolo_confidence"] = round(yolo_conf, 3)
    result["photo_crop_size"] = photo_crop.size

    return result


print("verify_face pipeline ready")
```
